```python
import numpy as np

def get_metrics(metric:str):
    if metric == 'euclidean':
        return lambda x, y: np.sqrt(np.sum((x - y) ** 2))
    elif metric == 'manhattan':
        return lambda x, y: np.sum(np.abs(x - y))
    elif metric == 'cosine':
        return lambda x, y: 1 - np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))
    else:
        raise ValueError(f"Unsupported metric: {metric}")
    

x = np.array([i for i in range(10)])

codebook = np.array([[i+j for i in range(1,11)] for j in range(10)])

metric = get_metrics('euclidean')
print(codebook)

c = np.argmax([metric(x, codebook[i]) for i in range(codebook.shape[0])])

print(codebook[c])
```
