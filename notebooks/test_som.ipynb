```python
from setup_ipynb_paths import setup_project_root
PROJECT_ROOT = setup_project_root()

from sklearn import datasets, preprocessing
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scripts.my_som import stepwiseSOM, batchSOM
from scripts.func4som import plot_som_2D
```

```python
# Génération de données synthétiques
X, y = datasets.make_blobs(n_samples=1000, centers=4, n_features=2, random_state=0)
scaler = preprocessing.StandardScaler()
X_scaled = scaler.fit_transform(X)

# Génération de données en forme de lunes
X_moons, y_moons = datasets.make_moons(n_samples=1000, noise=0.1, random_state=0)
X_moons_scaled = scaler.fit_transform(X_moons)

```

## **Stepwise SOM**

```python
som_stepwise_blobs = stepwiseSOM(0.9, 1, (5,5), 1000, 'pca')
codebook = som_stepwise_blobs.train(X_scaled)
plot_som_2D(codebook, X_scaled, y, title="SOM Stepwise - grille et codebook")
```

```python

som_stepwise_moons = stepwiseSOM(0.9, 1, (5,5), 1000, 'pca')
codebook_moons = som_stepwise_moons.train(X_moons_scaled)
plot_som_2D(codebook_moons, X_moons_scaled, y_moons, title="SOM Stepwise - Moons")
```

## **Batch SOM**

```python
batch_som_blobs = batchSOM(1, (5,5), 1000, 'pca')

cb_b = batch_som_blobs.train(X_scaled)
plot_som_2D(cb_b, X_scaled, y, title="SOM Batch - Codebook Vectors")
```

```python
som_batch_moons = batchSOM(1, (5,5), 1000, 'pca')
codebook_moons = som_batch_moons.train(X_moons_scaled)
plot_som_2D(codebook_moons, X_moons_scaled, y_moons, title="SOM Batch - Moons")
```
